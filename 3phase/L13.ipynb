### step 2 : Preprocessing the data

```python
import re # it is used for regular expression operations
def clean(text):
    text =  str(text).lower()
    text = re.sub(r'[^a-zA-Z0-9\s]', '', text) # keep letter space digits
    return text.strip().split() # return list of words

question = [clean(q) for q in df['question']]
answer = [clean(a) for a in df['answer']]
```

### step 3 : building vocabulary

```python
PAD, SOS, EOS, UNK = 0, 1, 2, 3
def built_vocab(sentences):
    vocab = {'<PAD>': PAD, '<SOS>': SOS, '<EOS>': EOS, '<UNK>': UNK}
    for sentence in sentences:
        for word in sentence:
            if word not in vocab:
                vocab[word] = len(vocab)
    return vocab

question_vocab = built_vocab(question)
answer_vocab = built_vocab(answer)    
```
